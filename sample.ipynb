```python
CONCEITOS = ["I", "R", "B", "MB"]

def classificar_valor(valor, regra):
    if regra == "-":
        return "-"
    if pd.isna(valor):
        return "SEM DADO"
    for limite, conceito in zip(regra, CONCEITOS):
        if valor <= limite:
            return f"{int(valor)}-{conceito}"
    return f"{int(valor)}-E"

def menor_mencao(mencao_completa):
    if any("-I" in v for v in mencao_completa):
        return "I"
    elif any("-R" in v for v in mencao_completa):
        return "R"
    elif any("-B" in v for v in mencao_completa):
        return "B"
    elif any("-MB" in v for v in mencao_completa):
        return "MB"
    elif any("-E" in v for v in mencao_completa):
        return "E"
    else:
        return "Dado invalido"

def suficiencia(lem, menor_conceito):

    if menor_conceito == "I" or menor_conceito == "R":
        return f"{lem}(NS)"

    elif menor_conceito == "B" and lem == "PED":
        return f"{lem}(NS)"

    else:
        return f"{lem}(S)"
```

```python
PROVAS = ['CORRIDA', 'FLEXÃO', 'ABDOMINAL', 'BARRA']

def formatar_regras(regras):
    if regras is None:
        return None
    for prova in PROVAS:
        regras.setdefault(prova, "-")
    return regras
```

```python
def avaliar_candidato(linha):
    pg = linha['P/G']
    nome_guerra = linha['NOME_DE_GUERRA']
    nome_completo = linha['NOME_COMPLETO']
    sexo = linha['SEXO']
    lem = linha['LEM']
    idade = linha['IDADE']

    regras = formatar_regras(buscar_regra(sexo, lem, idade))

    if regras is None:
        return pd.Series({
            "P/G": f"{pg}",
            "NOME_DE_GUERRA": f"{nome_guerra}",
            "NOME_COMPLETO": f"{nome_completo}",
            "IDADE": "-",
            "MENÇÃO COMPLETA": "-;",
            "M_MENCAO": "-;",
        })

    c_corrida = classificar_valor(linha.get("CORRIDA"), regras["CORRIDA"])
    c_flexao = classificar_valor(linha.get("FLEXÃO"), regras["FLEXÃO"])
    c_abdominal = classificar_valor(linha.get("ABDOMINAL"), regras["ABDOMINAL"])
    c_barra = classificar_valor(linha.get("BARRA"), regras["BARRA"])

    mencao_completa = [c_corrida, c_flexao, c_abdominal, c_barra]
    m_mencao = menor_mencao(mencao_completa)
    s_mencao = suficiencia(lem, m_mencao)


    return pd.Series({
        "P/G": f"{pg}",
        "NOME_DE_GUERRA": f"{nome_guerra}",
        "NOME_COMPLETO": f"{nome_completo}",
        "IDADE": f"{int(idade)}",
        "MENÇÃO COMPLETA": f"{c_corrida}; {c_flexao}; {c_abdominal}; {c_barra}; {m_mencao}; {s_mencao}",
        "M_MENCAO": m_mencao,
    })
```

```python
resultado = df.apply(avaliar_candidato, axis=1)

ORDEM_NOTA = ["E", "MB", "B", "R", "I"]
ORDEM_PG = ["Cap", "1º Ten", "2º Ten", "Asp Of", "S Ten",
            "1º Sgt", "2º Sgt", "3º Sgt", "Cb", "Sd EP", "Sd EV"]

resultado["M_MENCAO"] = pd.Categorical(resultado["M_MENCAO"], categories=ORDEM_NOTA, ordered=True)
resultado["P/G"] = pd.Categorical(resultado["P/G"], categories=ORDEM_PG, ordered=True)

resultado = resultado.sort_values(["M_MENCAO", "P/G"]).drop(columns="M_MENCAO")
```
